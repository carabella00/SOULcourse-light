### Part 1: Earth's Temperature
Since the Earth gets most of its heat from the Sun, we can calculate the temperature that Earth should be given the amount of light output by the Sun, our distance from the Sun, and the amount of that light that gets reflected away from Earth. Most light that reaches the Earth and does not get reflected away is absorbed as heat.

```python
luminosity = 3.83e26 # luminosity of the Sun in Watts
albedo = 0.3 # albedo of the Earth (fraction of light that gets reflected)
distance = 1.496e11 # distance of Earth from Sun in meters
sigma = 5.67e-8 # Stefan-Boltzmann constant
```

```python
T_eq = ( (luminosity * (1-albedo)) / (16 * np.pi * sigma * distance**2) )**(1/4)
print(f'Earth\'s Equilibrium Temperature: {round(T_eq)} K')
```

```python
T_eq_C = T_eq - 273.15
T_eq_F = ( T_eq - 273.15 ) * (9/5) + 32
print(f'in Celsius: {round(T_eq_C)} ºC')
print(f'in Fahrenheit: {round(T_eq_F)} ºF')
```

This seems... really cold. We might get these temperatures in the winter or in certain parts of the world, but the average surface temperature is considerably warmer at 288 K (15 ºC; 59 ºF). Why is the Earth's surface temperature so much warmer than its equilibrium temperature? What factors does equilibrium temperature ignore?

This effect is something we can model! Most atmospheric models split the atmosphere into layers, calculate the effects of small changes to each layer on the other layers, and iterate until a steady solution is reached. The model below is much more simple, using an approximate relationship between CO$_2$ concentration, heat forcing, and equilibrium warming to extrapolate concentrations. Of course, proper climate modeling would take all greenhouse gases into effect, but we just want to explore the effects of changing CO$_2$ only.

Change the values in the array in the first line to see how changing the abundance of CO$_2$ influences the heat forcing, warming, and subsequent surface temperature. For reference, today's CO$_2$ concentration of 432 ppm corresponds to a surface temperature of 288 K.

```python
CO2_ppm = np.array([0,0,0,0,0]) # change these!

# reference to today
CO2_ref = 432 # reference CO2 abundance (ppm)
T_ref = 288 # reference surface temperature (K)

# climate model priors
delta_F_ref = 5.35 * np.log(2) # added heat flux for doubling CO2 (from observations)
warming_ref = 3 # estimated temperature increase for doubling CO2
warming_factor = delta_F_ref / 3 # factor linking added heat flux to temp increase

# calculate change in radiative forcing
delta_F = 5.35 * np.log( CO2_ppm / CO2_ref )

# calculate change in temperature
delta_T = delta_F / warming_factor

# calculate new surface temperature
T_greenhouse = T_ref + delta_T

print('CO$_2$ (ppm) \t Forcing (W/m$^2$) \t Warming (K) \t Surface Temperature (K)')
for i in range(len(CO2_ppm)):
  print(f'{CO2_ppm[i]} \t\t {round(delta_F[i],1)} \t\t\t {round(delta_T[i],1)} \t\t {round(T_greenhouse[i],1)}')
```

We can plot this to make the trend easier to pull out. Let's plot the surface temperature as a function of CO$_2$ concentration.

```python
# define variables to plot
x = np.linspace(0,2000,1000)
y = 5.35 * np.log( x / CO2_ref ) / warming_factor + T_ref

# set up the plot
fig, ax = plt.subplots()
ax.plot(x, y,color='purple')
ax.scatter(CO2_ppm,T_greenhouse,color='gold',marker='*',s=100,zorder=2,label='your calculated values')

ax.axhline(288,color='gray')
ax.text(0.05,288.5,'Current day Earth (288 K)')

ax.set_xlabel('CO$_2$ Concentration (ppm)')
ax.set_ylabel('Surface Temperature (K)')
ax.legend(frameon=False,loc='lower right')
```

Looking at this plot, we see that Earth's surface temperature could vary from 260 to 295 K depending on CO$_2$ concentration. These are huge swings for global average temperature. Run the next cell to see the temperatures where things start to go wrong for life here on Earth.

```python
# define our variables to plot
x = np.linspace(0,2000,1000)
y = 5.35 * np.log( x / CO2_ref ) / warming_factor + T_ref

# set up the plot
fig,ax = plt.subplots()
ax.plot(x,y,color='purple',zorder=9)
ax.scatter(CO2_ppm,T_greenhouse,color='gold',marker='*',s=200,zorder=10)

# label the plot
ax.set_xlabel('CO$_2$ Concentration (ppm)')
# ax.set_xlim(0,1000)
ax.set_ylabel('Surface Temperature (K)')
ax.set_ylim(283,291)
ax.axhline(288,color='gray')
ax.text(550,288.1,'Current day Earth (288 K)')
ax.axhline(286.5,color='black')
ax.text(380,286.7,'Pre-industrial Earth (286 K)')

# doomsday scenarios
ax.axhline(286.5+1.0,color='gold')
ax.text(480,286.5+1.1,'Extreme weather events (+1 K)')
ax.axhline(286.5+3.0,color='red')
ax.text(800,286.5+3.4,'Catastrophic ice melt and')
ax.text(800,286.5+3.1,'sea level rise (+3 K)')
ax.axhline(286.5-1.0,color='turquoise')
ax.text(300,286.5-0.9,'Agricultural collapse (-1 K)')
ax.axhline(286.5-3.0,color='dodgerblue')
ax.text(200,286.5-2.9,'Ice age temperatures (-3 K)')
```

### Part 2: Radiation at Earth's Surface

The greenhouse effect is most prevalent in the thermal infrared, but as we saw from the Sun's blackbody curve and irradiance spectrum, there are lots of other wavelengths that make it to Earth. Here are standard solar irradiance spectra from The National Laboratory of the Rockies (original link https://www.nlr.gov/grid/solar-resource/spectra-am1.5), one from above Earth's atmosphere and one from the surface. These spectra go from the UV to the near infrared.

```python
# grab spectra from github
solar_spectra = pd.read_csv('https://raw.githubusercontent.com/carabella00/SOULcourse-light/main/ASTMG173.csv',sep=r",",skiprows=1)
# quick smoothing to make features easier to see
for i in ['Etr W*m-2*nm-1','Direct+circumsolar W*m-2*nm-1']:
  series = pd.Series(solar_spectra[i])
  solar_spectra[i] = series.rolling(window=5, min_periods=1).mean()

# set up the plot
fig,ax = plt.subplots(figsize=(12,5))
ax.plot(solar_spectra['Wvlgth nm']*1e-3,solar_spectra['Etr W*m-2*nm-1']*1e3,color='midnightblue',label='Top of atmosphere')
ax.plot(solar_spectra['Wvlgth nm']*1e-3,solar_spectra['Direct+circumsolar W*m-2*nm-1']*1e3,color='deepskyblue',label='At surface')

# label the plot
ax.set_xlabel('Wavelength (µm)')
ax.set_xlim(0.27,4)
ax.set_ylabel('Irradiance (W m$^{-2}$ µm$^{-1}$)')
ax.legend(frameon=False,fontsize=16)
```

There are clear differences between the spectrum at the top of the atmosphere and at the surface. These spectra are information-rich, telling us things like what gases are in the atmosphere and which wavelengths make it to the surface. The cell below will zoom into the three regions of the electromagnetic spectrum present in this plot and label some important wavelengths.

```python

# define wavelength ranges in µm
regions = [
    ('Ultraviolet', 0.27, 0.4),
    ('Visible', 0.4, 0.7),
    ('Infrared', 0.7, 4.0)
]

# plot the spectra in each region
fig,axs = plt.subplots(1,3,figsize=(16,5),sharey=True,gridspec_kw={'wspace':0.1, 'width_ratios':[1,1,1.5]})
for ax, (name, xmin, xmax) in zip(axs, regions):

    ax.plot(solar_spectra['Wvlgth nm']*1e-3,
            solar_spectra['Etr W*m-2*nm-1']*1e3,
            color='midnightblue', label='Top of atmosphere')

    ax.plot(solar_spectra['Wvlgth nm']*1e-3,
            solar_spectra['Direct+circumsolar W*m-2*nm-1']*1e3,
            color='deepskyblue', label='At surface')

    ax.set_xlim(xmin, xmax)
    ax.set_title(name, fontsize=14)

# label the plot
axs[0].set_ylabel('Irradiance (W m$^{-2}$ µm$^{-1}$)')
fig.supxlabel('Wavelength (µm)')
axs[2].legend(frameon=False, fontsize=12)
axs[0].axvline(0.315,color='gray',linestyle='--')
axs[0].text(0.29,1700,'UV-B',size=14)
axs[0].text(0.32,1700,'UV-A',size=14)
axs[2].axvspan(0.74,0.78,color='gray')
axs[2].text(0.78,1500,'O$_2$',size=14)
axs[2].axvspan(0.9,1.0,color='darkgray')
axs[2].axvspan(1.1,1.2,color='darkgray')
axs[2].axvspan(1.3,1.5,color='darkgray')
axs[2].axvspan(1.8,2.0,color='darkgray')
axs[2].text(1,1200,'H$_2$O features',size=14)
axs[2].axvspan(2.4,3.0,color='lightgray')
axs[2].text(2.5,900,'H$_2$O & CO$_2$',size=14)
axs[2].text(2.3,700,'overlapping feature',size=14)
```
