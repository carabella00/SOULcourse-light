```python
import matplotlib.pyplot as plt # plotting package
import numpy as np # numerical computation package
import pandas as pd # dataset manipulation package
```

We now have a bunch of tools to work with and plot data. Next, let's make our first plot.

In class, we talked about the Sun and the blackbody radiation it puts out...

```python
# constants (SI units)
h = 6.626e-34 # Planck's constant (J*s)
c = 3.0e8 # speed of light (m/s)
k = 1.381e-23 # Boltzmann constant (J/K)

# temperature of Sun (K)
T = 5800

# wavelengths (0.2 to 3 µm)
wl = np.linspace(.2,3,1000)
wl_m = wl * 1e-6 # convert µm to m

# Planck's law
intensity = (2 * h * c**2 / wl_m**5) / (np.exp(h * c / (wl_m * k * T)) - 1)
```

```python
plt.plot(wl,intensity,linewidth=5,label=f"T = {T} K")

plt.xlabel('Wavelength (µm)')
plt.ylabel('Intensity (W m$^{-3}$ sr$^{-1}$)')
plt.legend(fontsize=14,frameon=False)
plt.grid(which='major',alpha=0.3)
```

You just plotted the blackbody of the Sun! At what wavelength does the Sun's radiation peak? What part of the electromagnetic spectrum is that wavelength in?

```python
# temperatures (K)
temps = [0,0,0] # change these!

for T in temps:
  intensity = (2 * h * c**2 / wl_m**5) / (np.exp(h * c / (wl_m * k * T)) - 1)
  # intensity = intensity / max(intensity)
  plt.plot(wl,intensity,linewidth=5,label=f"T = {T} K")

plt.xlabel('Wavelength (µm)')
plt.ylabel('Intensity (W m$^{-3}$ sr$^{-1}$)')
plt.legend(fontsize=14,frameon=False)
plt.grid(which='major',alpha=0.3)
```

Okay, one last thing. So far, we plotted theoretical blackbody spectra using Planck's law. But how well does the model describe the actual Sun? Let's grab some real data. Here is the 1985 Wehrli standard Solar irradiance spectrum (originally from https://www.nlr.gov/grid/solar-resource/spectra-wehrli).

```python
# grab spectrum from github and print the variable
solar_spectrum = pd.read_csv('https://raw.githubusercontent.com/carabella00/SOULcourse-light/main/wehrli85.txt',sep=r"\s+")
print(solar_spectrum)
```

```python
plt.figure(figsize=(14, 5))

# plot blackbody of Sun at Earth (some extra math to get the value at Earth)
intensity = (2 * h * c**2 / wl_m**5) / (np.exp(h * c / (wl_m * k * 5800)) - 1) # in W m^-3 sr^-1
R_sun = 6.957e8 # radius of the Sun (m)
a = 1.496e11 # Earth-Sun distance (m)
bb_irradiance = intensity * np.pi * (R_sun / a)**2 * 1e-6
plt.plot(wl,bb_irradiance,color='goldenrod',linewidth=5,label='5800 K Blackbody')

# plot observed Solar spectrum
solar_spectrum['um'] = solar_spectrum['nm']*1e-3 # convert to µm
solar_spectrum['W/sm/um'] = solar_spectrum['W/sm/nm']*1e3 # convert to W/sm/µm
plt.plot(solar_spectrum['um'],solar_spectrum['W/sm/um'],color='black',label='Observed Sun')

plt.xlabel('Wavelength (µm)')
plt.xlim(0,3)
plt.ylabel('Irradiance (W m$^{-2}$ µm$^{-1}$)')
plt.legend(fontsize=14,frameon=False)
plt.grid(which='major',alpha=0.3)
```
